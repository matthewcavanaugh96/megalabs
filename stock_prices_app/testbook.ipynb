```python
import pandas as pd

df = pd.read_csv('daily_cleaned_price_data.csv')

df2 = pd.read_csv('daily_cleaned_metadata.csv')

df3 = pd.merge(df, df2, on="ticker")
```

```python
df3['close_pct_of_day1'] = (df3.groupby('ticker')['close'].transform(lambda x: (x / x.iloc[0]) * 100))
df3
```

```python
df4 = df3[['timestamp', 'ticker', 'name', 'close', 'close_pct_of_day1']]
df4
```

```python
df4['close_pct_of_day1'] = df4['close_pct_of_day1'].round(2)
```

```python
df4.to_parquet('daily_cleaned_price_data.parquet', index=False)
```

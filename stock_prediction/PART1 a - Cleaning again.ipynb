```python
cutoff_date = pd.to_datetime("2022-06-09").date()

# Keep FB through 2022-06-08
full_stock_df = full_stock_df[
    ~((full_stock_df["ticker"] == "FB") & (full_stock_df["timestamp"] >= cutoff_date))
]

# Keep META from 2022-06-09 onward
full_stock_df = full_stock_df[
    ~((full_stock_df["ticker"] == "META") & (full_stock_df["timestamp"] < cutoff_date))
]

full_stock_df["ticker"] = full_stock_df["ticker"].replace({"FB": "META"})

full_stock_df
```

```python
full_stock_df = full_stock_df.sort_values(by=['ticker', 'timestamp'])
```
